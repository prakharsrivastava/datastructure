```python
def generate_tree(func, *args):
    """Generates a recursion tree in a structured graphical format."""

    def trace(func, *args, depth=0, prefix="Root"):
        """Recursive tracer to capture function calls and print the tree."""

        # Convert args to a readable format
        display_args = ", ".join(map(str, args))

        # Print the current node with indentation for tree structure
        print(" " * (6 * depth) + f"{prefix} -> [{display_args}]")

        # Define a wrapper function for recursive calls (left & right)
        def wrapper(*inner_args, direction="L"):
            trace(func, *inner_args, depth=depth + 1, prefix=f"{direction} -> {inner_args[1] if len(inner_args) > 1 else 'End'}")

        # Call the function and pass our wrapper
        func(*args, callback=wrapper)

    # Start tracing from the root
    trace(func, *args, depth=0, prefix="Root")


# ---------------------- Binary String Generator with Recursion Tree ----------------------
def generate_binary(n, s="", callback=None):
    """Generates all binary strings of length n using backtracking."""

    if len(s) == n:  # Base case: when string reaches desired length
        print(s)
        return

    # Left branch: Add '0'
    callback(n, s + "0")

    # Right branch: Add '1'
    callback(n, s + "1")


def subsets(nums, index=0, path=[], callback=None):
    if index == len(nums):
        print(path)
        return

    if callback:
        # Include current number
        callback(nums, index + 1, path + [nums[index]])

        # Exclude current number
        callback(nums, index + 1, path+ [nums[index]])

# Generate Tree for Subsets
generate_tree(subsets, [1, 2])

# ---------------------- Generate Tree for Binary String Generation ----------------------
#generate_tree(generate_binary, 2, "")

```
